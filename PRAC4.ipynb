### 1. Univariate Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = df.columns

for col in numeric_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(7, 3))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)
    plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.countplot(x='quality', data=df)
plt.title('Distribution of Wine Quality')
plt.show()
```

```python
df['quality'].value_counts().sort_index()
```

### 2. Bivariate Analysis

```python
sns.boxplot(x='quality', y='alcohol', data=df)
plt.title('Alcohol vs Wine Quality')
plt.show()
```

```python
features = ['chlorides', 'citric acid', 'density', 'fixed acidity', 'free sulfur dioxide', 'pH', 'residual sugar', 'sulphates', 'total sulfur dioxide', 'volatile acidity']
plt.Figure(figsize=(7, 4))
for col in features:
    plt.figure(figsize=(7, 4))
    sns.boxplot(x='quality', y=col, data=df)
    plt.title(f'{col} vs Wine Quality')
    plt.show()
plt.title(f'{col} vs Wine Quality')
plt.show()
```

```python
sns.scatterplot(x='alcohol', y='quality', data=df)
plt.title('Alcohol vs Wine Quality')
plt.show()
```

```python
sns.scatterplot(x='volatile acidity', y='quality', data=df)
plt.title('Volatile Acidity vs Wine Quality')
plt.show()
```

### 3.Multiversal Analysis

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Correlation Heatmap')
plt.show()
```

```python
#Pair Plot
selected = ['alcohol', 'volatile acidity', 'citric acid', 'residual sugar', 'density', 'quality']
sns.pairplot(df[selected], hue='quality')
plt.show()
```
